```python
import pandas as pd
import geopandas as gpd
from shapely.geometry import Point
```

```python
df = pd.read_csv('ChicagoCrimes2026.csv')
df.info()
```

- using long/lat data coordinates to get neighborhood names
    - need to do a spatial join: basically matching data based on geographic location (compares geometry meaning it checks whether one shape falls inside another shape)
    - chicago is broken into 77 polygons 
    - trying to check if this (x,y) fall inside any of these 77 polygons

- using chicago city boundaries geofile
    - geofile: data file that stores info abt geographic locations, physical boundaries, and spatial features



```python
geometry = [Point(xy) for xy in zip(df.Longitude, df.Latitude)]
points = gpd.GeoDataFrame(df, geometry = geometry, crs = "EPSG: 4326")

areas = gpd.read_file("ChicagoBoundaries.geojson")

result = gpd.sjoin(points, areas, how = "left", predicate = "within")
```
